# Basic Analysis of the Dataset:

```python
import pandas as pd
import matplotlib as plt
import seaborn as sns
```

### Load in data and check target:

```python
# This alone will take a long time

df = pd.read_csv("../Datasets/trans_3000p2_list.csv", low_memory=True) # need to rename original IBM download to .csv
```

```python
# Basic info
df.info(memory_usage='deep')

# Target Balance
target_share = df['Is Laundering'].value_counts()
print(f"Total laundering: {target_share.get(1, 0)}")
print(f"Total clean: {target_share.get(0, 0)}")

fraud_rate = (target_share.get(1, 0) / target_share.sum()) * 100
print(f"{fraud_rate:.4f}% laundering Rate\n")

# Total accounts 

num_unique_from_accounts = df['Account'].nunique()
num_unique_to_accounts = df['To Account'].nunique()

print(f"Unique From Accounts: {num_unique_from_accounts}")
print(f"Unique To Accounts: {num_unique_to_accounts}")

# Total unique entities across both sides 
unique_all_accounts = pd.unique(df[['Account', 'To Account']].values.ravel('K'))
print(f"Total unique accounts (combined): {len(unique_all_accounts)}")

```
